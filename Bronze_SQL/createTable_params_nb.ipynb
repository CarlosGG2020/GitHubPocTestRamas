```python
dbutils.widgets.text("p_catalog", "")
dbutils.widgets.text("p_schema", "")

catalogo = dbutils.widgets.get("p_catalog")
esquema = dbutils.widgets.get("p_schema")

query = f"""
CREATE OR REPLACE TABLE `{catalogo}`.`{esquema}`.`tabla_dinamica_creada_airports` AS
SELECT
    *
FROM `{catalogo}`.`{esquema}`.`airports`

"""

spark.sql(query)
   
```
